Create Constructor

Create Railway Reservation Class using Constructor

```python
class RailwayReservation:

    def __init__(
        self,
        passenger_name,
        train_name,
        seat_number
    ):

        self.passenger_name = passenger_name
        self.train_name = train_name
        self.seat_number = seat_number
```

Create reservation object

```python
ticket1 = RailwayReservation(
    "Rahul Sharma",
    "Rajdhani Express",
    "A1-21")
```

Display reservation details

```python
print(ticket1.passenger_name)
print(ticket1.train_name)
print(ticket1.seat_number)
```

Handle Multiple Reservation Object

```python
class Reservation:

    def __init__(
        self,
        passenger_name,
        seat_number
    ):
        self.passenger_name = passenger_name
        self.seat_number = seat_number

    def display_ticket(self):
        print("Passenger :", self.passenger_name)
        print("Seat Number :", self.seat_number)
        print("---------------------------")
```

Create multiple reservation objects

```python
ticket1 = Reservation(
    "Rahul Sharma",
    "B1-21")

ticket2 = Reservation(
    "Priya Singh",
    "B1-22")

ticket3 = Reservation(
    "Amit Verma",
    "B1-23")
```

Display reservation details

```python
ticket1.display_ticket()
ticket2.display_ticket()
ticket3.display_ticket()
```

Build Final System & Use Destructor

Create Railway Reservation Class

```python
class RailwayReservation:

    def __init__(
        self,
        passenger_name,
        train_name,
        seat_number,
        ticket_price):

        self.passenger_name = passenger_name
        self.train_name = train_name
        self.seat_number = seat_number
        self.ticket_price = ticket_price
        print("Reservation Created Successfully")

    def display_ticket(self):

        print("===================================")
        print(" Railway Reservation Ticket ")
        print("===================================")
        print("Passenger Name :", self.passenger_name)
        print("Train Name :", self.train_name)
        print("Seat Number :", self.seat_number)
        print("Ticket Price :", self.ticket_price)

    def __del__(self):
        print("Reservation Cancelled")
```

Create Object

```python
ticket1 = RailwayReservation(
    "Neha Kapoor",
    "Rajdhani Express",
    "A1-25",
    2450)
```

Display reservation ticket

```python
ticket1.display_ticket()
```

Delete Reservation Object

```python
del ticket1
```
