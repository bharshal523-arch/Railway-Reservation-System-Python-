Create Lambda Function for Fare Discount

Create lambda function

```python
discount_fare = lambda fare: fare - 200
```

Print Result

```python
print("Discounted Fare :", discount_fare(2500))
```

Creation of List

```python
ticket_fare = [1600, 2000, 2500, 3000, 3500]
```

Apply map() Function for Fare Calculation



Apply map() function

```python
updated_fares = list(

   map(
       lambda fare: fare + 100,
       ticket_fare))
```

Display updated fares

```python
print(updated_fares)
```

Apply filter() Function for confirmed tickets



Create reservation status list.

```python
reservation_status = [
   "Confirmed",
   "Waiting",
   "Confirmed",
   "RAC"]
```

Apply filter() function

```python
confirmed_tickets = list(
   filter(
       lambda status: status == "Confirmed",
       reservation_status))
```

Print Output

```python
print(confirmed_tickets)
```

Process Data Using map() and filter()

Create fare list

```python
ticket_fares = [2500, 800, 3200, 600]
```

Filter valid fares

```python
valid_fares = list(
   filter(
       lambda fare: fare > 1000,
       ticket_fares))
```

Apply fare update

```python
final_fares = list(
   map(
       lambda fare: fare + 100,
       valid_fares))
```

Output

```python
print(final_fares)
```

Build Final Functional Reservation Module

Create reservation list

```python
passengers = ["Rahul", "Priya", "Amit"]
seats = ["S1", "S2", "S3"]
fares = [2500, 1800, 3200]
```

Combine reservation data

```python
reservation_data = list(
   zip(passengers, seats, fares))
```

Display Final Reservation Report

```python
print("===================================")
print(" Railway Reservation Report ")
print("===================================")

for index, data in enumerate(reservation_data):

   passenger, seat, fare = data

   final_fare = (lambda x: x - 100)(fare)

   print("Reservation ID :", index + 1)
   print("Passenger :", passenger)
   print("Seat :", seat)
   print("Final Fare :", final_fare)
   print("-----------------------------------")
```
