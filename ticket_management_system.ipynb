Perform Ticket Fare Calculations

1) Create variables for passenger count and fare calculation

```python
number_of_passengers = 3
ticket_price = 2450
```

2) Calculate total ticket amount

```python
total_fare = number_of_passengers * ticket_price
print("Total Fare :", total_fare)
```

 Calculate GST on Ticket Fare
 1) Create a GST percentage variable

```python
gst_percentage = 5
```

2) Calculate GST amount

```python
gst_amount = (total_fare * gst_percentage) / 100
print("GST Amount :", gst_amount)
```

3) Calculate final payable amount

```python
final_amount = total_fare + gst_amount
print("Final Payable Amount :", final_amount)
```
