Create Function Using Default Arguments



Define reservation function

```python
def reservation_status(
   passenger_name,
   ticket_status="Confirmed"):
   print("Passenger Name :", passenger_name)
   print("Ticket Status :", ticket_status)
```

Call the function



```python
reservation_status("Rahul Sharma")
reservation_status("Amit Verma", "Waiting")
```

Use *args (Variable Length Arguments)

Create a function to handle multiple passenger names

```python
def passenger_booking(*passengers):
   print("Passenger List:")
   for name in passengers:
       print("Passenger :", name)
```

Call the function with multiple variables

```python
passenger_booking(
   "Rahul Sharma",
   "Priya Singh",
   "Amit Verma",
   "Neha Kapoor")
```

 Use **kwargs (Keyword Variable Arguments)

Create a function to store passenger ticket details dynamically

```python
def ticket_details(**details):
   print("Ticket Information:")
   for key, value in details.items():
       print(key, ":", value)
```

Call the function

```python
ticket_details(
   passenger_name="Rahul Sharma",
   train_name="Rajdhani Express",
   seat_number="B1-21",
   fare=2500,
   status="Confirmed")
```

Create Recursive Seat Countdown Function

Define recursive function

```python
def seat_countdown(seats):
   if seats == 0:
       print("All Seats Booked")
       return
   print("Remaining Seats :", seats)
   seat_countdown(seats - 1)
```

Call the recursive function

```python
seat_countdown(5)
```

Create Recursive Passenger Verification Function

Define recursive verification function

```python
def verify_passenger(passenger_number):
   if passenger_number == 0:
       print("Verification Completed")
       return
   print("Verifying Passenger :", passenger_number)
   verify_passenger(passenger_number - 1)
```

Call the function

```python
verify_passenger(3)
```

Build Final Advanced Booking Module

Create final booking function

```python
def final_booking(
   train_name="Rajdhani Express",
   *seat_numbers
):
   print("===================================")
   print(" Railway Reservation ")
   print("===================================")
   print("Train Name :", train_name)
   for seat in seat_numbers:
       print("Allocated Seat :", seat)
```

Call the function

```python
final_booking(
   "Rajdhani Express",
   "B1-21",
   "B1-22",
   "B1-23")
```
